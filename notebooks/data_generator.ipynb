```python
import pandas as pd 
import numpy as np 

np.random.seed(42)

n=50000

data=pd.DataFrame({
    "transaction_id": np.arange(1, n+1),
    "user_id": np.random.randint(1000, 9999, n),
    "amount": np.random.exponential(scale=800, size=n).round(2),
    "transaction_type": np.random.choice(["P2P", "P2M", "QR", "Autopay"], n),
    "merchant_category": np.random.choice(["Food", "Travel", "Shopping", "Bills", "Other"], n),
    "device_type": np.random.choice(["Android", "iOS"], n),
    "time_hour": np.random.randint(0, 24, n),
    "location_match": np.random.choice([0, 1], n, p=[0.2, 0.8]),
    "num_txn_last_24h": np.random.poisson(3, n),
    "avg_amount_last_7d": np.random.normal(1200, 300, n).round(2),
    "is_blacklisted_merchant": np.random.choice([0, 1], n, p=[0.95, 0.05]),
})

# Fraud logic
data["fraud"] = (
    (data["amount"] > 20000).astype(int) |
    ((data["location_match"] == 0) & (data["amount"] > 5000)).astype(int) |
    ((data["num_txn_last_24h"] > 15)).astype(int) |
    (data["is_blacklisted_merchant"] == 1).astype(int)
)

data["fraud"] = np.where(np.random.rand(n) < 0.01, 1, data["fraud"])  # random noise

data.to_csv("upi_transactions.csv",index=False)

print("Dataset generated: upi_transactions.csv")
```
